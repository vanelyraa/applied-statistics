# Problem 1: Exploring Petal Length

1. Plot a histogram showing the petal lengths of all 150 flowers.
2. On the same plot, superimpose an appropriate normal distribution, using the mean and standard deviation calculated from the data.

```python
# Reading Iris dataset
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")
```

```python
df
```

```python
# Variable storing petal lenghts
petal = df["petal_length"]
```

```python
# Finding min and max petal lengths
petal_min = petal.min()
petal_max = petal.max()

# Showing min and max values
petal_min, petal_max
```

```python
# Calculating mean and standard deviation
petal_mean = petal.mean()
petal_std = petal.std()

# Showing values
petal_mean,petal_std
```

```python
# Creating histogram
fig,ax = plt.subplots(figsize=(10,5))

# Histogram of all 150 petal lengths
ax.hist(petal, edgecolor="black",density=True, bins=np.linspace(1.0, 7.0, 50))

# Normal distribution x values
x = np.linspace(petal_min,petal_max, 50)

# Normal distribution plot
ax.plot(x, stats.norm.pdf(x,petal_mean,petal_std))

# Plot labels
ax.set_xlabel("Petal length (cm)")
ax.set_ylabel("Density")
ax.set_title("Petal length of all 150 flowers");
```

# Problem 2: Correlation Between Petal Length and Petal Width

Investigating the relationship between petal_length and petal_width in the Iris dataset.
1.	Calculate the Pearson correlation coefficient between petal length and petal width using all 150 flowers	

```python
# Pearson correlation for all 150 flowers
petal_pcorrelation = df["petal_length"].corr(df["petal_width"])

# Showing correlation
petal_pcorrelation
```

https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.corr.html: correlation and dataframe 

2. Create an appropriate scatter plot showing the relationship between the two variables, and include a suitable line to help illustrate the relationship.


```python
# Length and width variables 
petal_length = df["petal_length"]
petal_width = df["petal_width"]

# Calculating slope and intercept
m, c = np.polyfit(petal_length, petal_width, 1)

# Fitted line values
x = np.linspace(petal_length.min(), petal_length.max(), 100)

# Create splot
fig, ax = plt.subplots(figsize=(10, 5))

ax.scatter(petal_length, petal_width, label="Iris flowers")

# Plot fitted straight line
ax.plot(x, m * x + c, color="red", label="Fitted line")

# Plot labels
ax.set_xlabel("Petal length (cm)")
ax.set_ylabel("Petal width (cm)")
ax.set_title("Petal Length and Petal Width")
ax.legend()

plt.show()
```
