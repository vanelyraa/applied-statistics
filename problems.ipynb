```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

# Problem 1: Exploring Petal Length

1. Plot a histogram showing the petal lengths of all 150 flowers.
2. On the same plot, superimpose an appropriate normal distribution, using the mean and standard deviation calculated from the data.

```python
# Reading Iris dataset
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")
```

```python
df
```

```python
# Variable storing petal lenghts
petal = df["petal_length"]
```

```python
# Finding min and max petal lengths
petal_min = petal.min()
petal_max = petal.max()

# Showing min and max values
petal_min, petal_max
```

```python
# Calculating mean and standard deviation
petal_mean = petal.mean()
petal_std = petal.std()

# Showing values
petal_mean,petal_std
```

```python
# Creating histogram
fig,ax = plt.subplots(figsize=(10,5))

# Histogram of all 150 petal lengths
ax.hist(petal, edgecolor="black",density=True, bins=np.linspace(1.0, 7.0, 150))

# Normal distribution x values
x = np.linspace(petal_min,petal_max, 150)

# Normal distribution plot
ax.plot(x, stats.norm.pdf(x,petal_mean,petal_std))

# Plot labels
ax.set_xlabel("Petal length (cm)")
ax.set_title("Petal length of all 150 flowers");
```
